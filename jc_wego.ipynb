# **Q1**

What is the overall on-time performance, and what do the overall distribution of adherence look like?

```python
working_table = wego.copy()
```

```python
avg_ontime = working_table['ADJUSTED_ONTIME_COUNT'].mean()
avg_ontime
```

```python
working_table['ON_TIME'] = np.where((working_table['ADHERENCE'].notnull()) & (working_table['ADHERENCE'].between(-6, 6)), 1, 0)
working_table
```

```python
working_table.to_csv('Workingtable.csv', index=False)
```

```python
total_count = working_table['ADHERENCE'].notnull().sum()
ontime_count = working_table['ON_TIME'].sum()
ontime_perc = (ontime_count / total_count) * 100
```

```python
print(f"Overall On-Time Performance: {ontime_perc:.2f}%")
```

# **Q4**

How much of a factor does the driver have on on-time performance? The driver is indicated by the OPERATOR variable.

```python
wego.columns
```

```python
drivers = wego.groupby('OPERATOR', as_index=False)[
    ['ADJUSTED_EARLY_COUNT', 'ADJUSTED_LATE_COUNT', 'ADJUSTED_ONTIME_COUNT']
].sum()

drivers
```

```python
# New column to calculate each operators adherence
drivers['TOTAL_COUNTS'] = (drivers['ADJUSTED_EARLY_COUNT'] + drivers['ADJUSTED_LATE_COUNT'] + drivers['ADJUSTED_ONTIME_COUNT'] )
```

```python
drivers['ONTIME_PERC'] = round((drivers['ADJUSTED_ONTIME_COUNT'] / drivers['TOTAL_COUNTS'] * 100), 2)
```

```python
drivers = drivers.sort_values('ONTIME_PERC', ascending=False)
```

```python
drivers = drivers.reset_index(drop=True)
```

```python
drivers
```

```python
drivers.to_csv('WeGo_drivers_dataset.csv', index=False)
```

```python
correlation = drivers.corr()
```

```python
correlation.to_csv('Driver_Correlation.csv', index=False)
```

```python
correlation = correlation.iloc[3:6, 3:6]
```

```python
bus_lightyear_colors = ["#512DA8", "#1976D2", "#FFFFFF", "#8BC34A"]

bus_lightyear_cmap = LinearSegmentedColormap.from_list("bus_lightyear", bus_lightyear_colors)
```

```python
correlation_styled = correlation.style.background_gradient(cmap=bus_lightyear_cmap) #adding a gradient based on values in cell

correlation_styled
```

```python
corr_long = (
    correlation
    .reset_index()                      # turn the row index into a column
    .melt(id_vars='index',              # melt all other columns
          var_name='Variable2',         # name for the column headers
          value_name='Correlation')     # name for the correlation values
    .rename(columns={'index': 'Variable1'})
)
```

```python
corr_long
```

```python
corr_long['Correlation'] = corr_long['Correlation'].round(3)
```

```python
corr_long.to_csv('correlation_long.csv', index=False)
```
